```python
import cv2
img=cv2.imread(r"opencv_image.png")
img=cv2.resize(img,(400,400))
```

```python
cv2.imshow("Image",img)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

```python
face_cascade=cv2.CascadeClassifier(r"haarcascade_frontalface_alt.xml")
```

```python
face_cascade
```

```python
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
cv2.imshow("Gray Image",img_gray)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

```python
face=face_cascade.detectMultiScale(img_gray,scaleFactor=1.1,minNeighbors=5)
```

```python
face
```

```python
for x,y,w,h in face:
    img=cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)
cv2.imshow("Image with faces",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
video=cv2.VideoCapture(0)#for web cam -video face detection
while True:
    ret,frame=video.read()
    if ret==False:
        break
    gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    face=face_cascade.detectMultiScale(gray,scaleFactor=1.1,minNeighbors=5)
    for x,y,w,h in face:
       img=cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),3)
    cv2.imshow("Video",frame)
    if cv2.waitKey(1) & 0xFF==ord("q"):
        break
video.release()
cv2.destroyAllWindows()
```
